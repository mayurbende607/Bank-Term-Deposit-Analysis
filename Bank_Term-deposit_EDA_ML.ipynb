# ML

# Data Preprocessing

```python
# Prepare Data
df['y'] = df['y'].map({'no':0,'yes':1})
df['y'].value_counts()
```

```python
# Separate Features and Target
X = df.drop('y', axis=1)
y = df['y']
```

```python
#Encode Categorical Variables
X = pd.get_dummies(X, drop_first=True)
```

```python
#Train-Test Split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print("Training Features :", X_train.shape)
print("Testing Features :", X_test.shape)

print("Training Target :", y_train.shape)
print("Testing Target :", y_test.shape)
```

#### Observation 
The dataset was split into training and testing sets using an 80:20 ratio.

After one-hot encoding, the number of input features increased from the original dataset to 53 features. The training dataset contains 32,950 records, while the testing dataset contains 8,238 records. This prepared dataset will be used to train and evaluate the machine learning models.

##  Lomistic Regression Model

```python
# Train Logistic Regression Model

# Import Logistic Regression
from sklearn.linear_model import LogisticRegression

# Create the model
lr_model = LogisticRegression(max_iter=1000, random_state=42)

# Train the model
lr_model.fit(X_train, y_train)
```

```python
# Make Predictions
# Predict on test data
y_pred_lr = lr_model.predict(X_test)

# Display first 10 predictions
print(y_pred_lr[:10])
```

```python
# Calculate Accuracy
from sklearn.metrics import accuracy_score

lr_accuracy = accuracy_score(y_test, y_pred_lr)

print(f"Logistic Regression Accuracy: {lr_accuracy:.2%}")
```

The Logistic Regression model achieved an accuracy of approximately 91%. This indicates that the model correctly classified most customer subscription outcomes on the test dataset.

The model can effectively predict whether a customer is likely to subscribe to a term deposit, helping the bank focus marketing efforts on customers with a higher probability of subscription.

```python
# Confusion Matrix  Evaluate
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred_lr)

print(cm)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred_lr))
```

#### Observation 
The Logistic Regression model achieved an accuracy of 91.07%. It performed very well in identifying customers who did not subscribe to a term deposit, achieving a recall of 97% for the "No" class. However, the model achieved a recall of only 40% for the "Yes" class, indicating that many actual subscribers were not identified. This suggests that while the model provides a strong baseline, it has limited ability to detect potential subscribers due to the class imbalance.

# Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier

# Create model
dt_model = DecisionTreeClassifier(
    random_state=42,
    max_depth=5
)

```

```python
# Train model
dt_model.fit(X_train, y_train)
```

```python
# Predictions
y_pred_dt = dt_model.predict(X_test)
```

```python
# Accuracy
dt_accuracy = accuracy_score(y_test, y_pred_dt)

print(f"Decision Tree Accuracy: {dt_accuracy:.2%}")
```

```python
# Evaluate
from sklearn.metrics import confusion_matrix

cm_dt = confusion_matrix(y_test, y_pred_dt)

print(cm_dt)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred_dt))
```

#### Observation
The Decision Tree model achieved an accuracy of 91.50%, slightly outperforming Logistic Regression (91.07%). It also improved the recall for the subscription class from 40% to 53%, meaning it identified more customers who were likely to subscribe to a term deposit. This makes the Decision Tree more effective for marketing campaigns where identifying potential subscribers is important.
    

# Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

# Create model
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

```

```python

# Train model
rf_model.fit(X_train, y_train)
```

```python
# Evaluate  Confusion Matrix
cm_rf = confusion_matrix(y_test, y_pred_rf)

print(cm_rf)
```

```python
print(classification_report(y_test, y_pred_rf))
```

#### Observation
Three machine learning models were developed and compared to predict term deposit subscriptions. Among them, the Decision Tree achieved the highest accuracy (91.50%) and the best recall for identifying customers likely to subscribe. Logistic Regression served as a strong baseline model, while Random Forest also produced competitive results but did not outperform the Decision Tree in this experiment.

# Feature Importance

```python
feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": rf_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

feature_importance.head(10)
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    data=feature_importance.head(10),
    x="Importance",
    y="Feature",
    palette="viridis"
)

plt.title("Top 10 Important Features")
plt.xlabel("Feature Importance")
plt.ylabel("Features")

plt.show()
```

## Observation

* Call duration was the most important feature in the model. However, duration is only known after the call, so it should not be used when the objective is to identify potential subscribers before contacting them..
- **Euribor 3-Month Rate (euribor3m)** is the second most important feature, indicating that economic conditions influence customer decisions.
- **Age** also plays a significant role, suggesting that customer demographics affect subscription behavior.
- **Number of Employees (nr.employed)** and **Campaign Contacts (campaign)** have moderate importance in the prediction.
- Features such as **pdays**, **cons.conf.idx**, **emp.var.rate**, **previous campaign outcome (poutcome_success)**, and **consumer price index (cons.price.idx)** also contribute to the model but with lower importance.

#### The duration feature is only known after the phone call has finished. Therefore, it cannot be used for predicting subscriptions before contacting a customer. If the goal is to identify customers before a campaign starts, the model should be retrained without the duration feature.

# Business Recommendations

1. **Prioritize Previous Campaign Success**
   - Customers with a successful previous campaign outcome showed the highest historical subscription rate. This segment can be prioritized for future campaigns.

2. **Use Customer Segmentation**
   - Subscription rates vary across age, job, education, and marital-status groups. Segmenting customers can help create more targeted campaigns.

3. **Optimize Contact Frequency**
   - Subscription rates generally decline as the number of campaign contacts increases. Future campaigns should monitor repeated contacts and avoid unnecessary follow-ups.

4. **Consider Campaign Timing**
   - Certain months show higher historical subscription rates than others. Campaign planning can consider these seasonal patterns.

5. **Prioritize Effective Contact Channels**
   - Cellular communication showed a higher historical subscription rate than telephone communication in this dataset.

6. **Use Previous Campaign Information**
   - Previous campaign outcomes provide valuable information for customer targeting and should be incorporated into predictive models.

7. **Use Machine Learning for Customer Prioritization**
   - The Decision Tree model can be used as a starting point for identifying customers with higher predicted subscription likelihood.

8. **Avoid Using Post-Contact Information for Pre-Campaign Prediction**
   - Call duration was highly important in the model, but it is only known after the call. Therefore, it should be excluded from a model intended for pre-campaign targeting.

9. **Monitor Model Performance**
   - Since the subscription class is relatively small, recall and F1-score should be monitored along with accuracy.

10. **Retrain with New Campaign Data**
   - The model should be periodically evaluated and retrained as new customer and campaign data become available.
